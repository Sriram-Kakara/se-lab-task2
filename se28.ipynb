```python
# Priority Scheduling (Preemptive)

n = int(input("Enter number of processes: "))
at, bt, pr = [], [], []

for i in range(n):
    at.append(int(input(f"AT P{i+1}: ")))
    bt.append(int(input(f"BT P{i+1}: ")))
    pr.append(int(input(f"Priority P{i+1}: ")))

rt = bt[:]
ct = [0]*n
time = 0
completed = 0

while completed < n:
    idx = -1
    min_pr = 999
    for i in range(n):
        if at[i] <= time and rt[i] > 0 and pr[i] < min_pr:
            min_pr = pr[i]
            idx = i
    time += 1
    rt[idx] -= 1
    if rt[idx] == 0:
        ct[idx] = time
        completed += 1

tat = [ct[i] - at[i] for i in range(n)]
wt = [tat[i] - bt[i] for i in range(n)]

print("WT:", wt)
print("TAT:", tat)
```
