```python
# Round Robin Scheduling

n = int(input("Enter number of processes: "))
bt = []

for i in range(n):
    bt.append(int(input(f"BT P{i+1}: ")))

qt = int(input("Enter Time Quantum: "))

rt = bt[:]
ct = [0]*n
time = 0

while sum(rt) > 0:
    for i in range(n):
        if rt[i] > 0:
            if rt[i] > qt:
                rt[i] -= qt
                time += qt
            else:
                time += rt[i]
                ct[i] = time
                rt[i] = 0

tat = ct
wt = [tat[i] - bt[i] for i in range(n)]

print("WT:", wt)
print("TAT:", tat)
```
