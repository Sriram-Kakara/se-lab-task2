```python
# First Come First Serve Scheduling

n = int(input("Enter number of processes: "))
bt = []

for i in range(n):
    bt.append(int(input(f"BT P{i+1}: ")))

wt = [0]*n
tat = [0]*n

for i in range(1, n):
    wt[i] = wt[i-1] + bt[i-1]

for i in range(n):
    tat[i] = wt[i] + bt[i]

print("WT:", wt)
print("TAT:", tat)
```
