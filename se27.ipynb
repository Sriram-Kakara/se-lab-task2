```python
#shell sort

arr = [5, 3, 8, 4, 2]

n = len(arr)
gap = n // 2

while gap > 0:
    for i in range(gap, n):
        temp = arr[i]
        j = i
        while j >= gap and arr[j - gap] > temp:
            arr[j] = arr[j - gap]
            j -= gap
        arr[j] = temp
    gap //= 2

print(arr)
```
