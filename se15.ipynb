```python
# Highest Response Ratio Next Scheduling

n = int(input("Enter number of processes: "))
at, bt = [], []

for i in range(n):
    at.append(int(input(f"AT P{i+1}: ")))
    bt.append(int(input(f"BT P{i+1}: ")))

wt = [0]*n
tat = [0]*n
completed = [False]*n
time = 0

for _ in range(n):
    idx = -1
    max_rr = -1
    for i in range(n):
        if at[i] <= time and not completed[i]:
            rr = (time - at[i] + bt[i]) / bt[i]
            if rr > max_rr:
                max_rr = rr
                idx = i
    wt[idx] = time - at[idx]
    time += bt[idx]
    tat[idx] = wt[idx] + bt[idx]
    completed[idx] = True

print("WT:", wt)
print("TAT:", tat)
```
